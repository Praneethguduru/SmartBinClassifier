```python
import pandas as pd
import numpy as np

OUT = "../data/prepared"
train = pd.read_csv(f"{OUT}/train/labels.csv", index_col=0)
val = pd.read_csv(f"{OUT}/validation/labels.csv", index_col=0)

# Baseline: for each product, always guess its most common count (this is 0)
guess = train.mode().iloc[0]
pred = pd.DataFrame([guess.values] * len(val), index=val.index, columns=val.columns)

print("Quantity exact-match accuracy:", round((pred == val).values.mean(), 3))
print("Average counting error (MAE):", round((pred - val).abs().values.mean(), 3))
print("Products it ever finds:", int((pred > 0).values.sum()))
```
